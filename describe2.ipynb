```python

# Step 1: Group and aggregate
agg_df = df.groupby(['Problem','Algo']).agg(
    count_run=('run','count'),
    min_val=('Optimal','min'),
    max_val=('Optimal','max'),
    mean_val=('Optimal','mean'),
    std_val=('Optimal','std')
)

# Step 2: Stack the metrics and pivot
final_df = agg_df.stack().unstack('Algo')

# Optional: Rename the index for clarity
final_df = final_df.rename_axis(['Problem','Metric']).reset_index()

# Display
print(final_df)
```

```python
final_df.to_csv(r"C:\Users\NOUHAILA\PycharmProjects\QLSA\resultTransposed_04012026.csv")
```

```python
# Set multi-index (Problem, Algo), stack metrics, then unstack Algo
reshaped = (
    df.set_index(["Problem", "Algo"])
      .stack()                # metrics become rows
      .reset_index()
      .rename(columns={"level_2": "Metric", 0: "Value"})
      .pivot_table(index=["Problem", "Metric"], columns="Algo", values="Value")
)

# Optional: reset index for a flat DataFrame
reshaped = reshaped.reset_index()

print(reshaped)
```
